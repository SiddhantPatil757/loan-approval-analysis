```python
import pandas as pd
import numpy as np

df = pd.read_csv("D:\Study\Codeneara\Project\Finace_Project\Loan_Approval_Rrisk_Pattern_Analysis\Data.CSV")

df.head()
```

```python
df.columns
df.info()
df.isnull().sum()
```

```python
df.columns = df.columns.str.strip()
```

```python
df['rejection_reason'] = df['rejection_reason'].fillna("Not Rejected")
```

```python
# Loan Approval Rate

df['loan_approval_status'].value_counts(normalize=True) * 100
```

```python
# CIBIL Score vs Loan Approval

pd.crosstab(df['risk_category'], df['loan_approval_status'])
```

```python
# Previous Default vs Approval

pd.crosstab(df['previous_default'], df['loan_approval_status'])
```

```python
# Debt-to-Income Ratio vs Approval

df.groupby('loan_approval_status')['debt_to_income_ratio'].mean()
```

```python
# EMI-to-Income Ratio vs Approval

df.groupby('loan_approval_status')['EMI_to_Income_Ratio'].mean()
```

```python
# Income vs Loan Amount

df[['annual_income', 'loan_amount']].corr()
```

```python
# Risk Category Distribution

df['risk_category'].value_counts()
```

```python
# Rejection Reasons

df[df['loan_approval_status'] == 'Rejected']['rejection_reason'].value_counts()
```

```python
# bank behavior differences

pd.crosstab(df['lender_type'], df['loan_approval_status'])
```
